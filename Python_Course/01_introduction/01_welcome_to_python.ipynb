## 2. Variables — labelled boxes for your data

Before a computer can *work* with a value, it needs somewhere to *keep* it. A **variable** is a name attached to a value — think of it as a **labelled box**:

```
┌──────────┐    ┌───────┐    ┌──────────┐
│  "Alice" │    │  25   │    │   1.70   │
└────┬─────┘    └───┬───┘    └────┬─────┘
     │              │             │
   name            age         height
```

You put a value into a box with the `=` sign (read it as *"gets"*, not *"equals"*): `age = 25` means *"the box labelled `age` gets the value 25"*. From then on, writing `age` anywhere means *"look inside that box"*.

````python
name = "Alice"      # text goes in quotes
age = 25            # a whole number
height = 1.70       # a decimal number

print(name)
print(age)
print(height)

# Variables can be reused and combined:
birth_year = 2026 - age
print(birth_year)
````

**Reading the output:** the first three lines simply show what is inside each box. One tiny surprise: we typed `1.70`, but Python printed `1.7` — Python stores the *value* of a number, not how you typed it. (Making numbers look nice for humans is a separate job, and Section 5 has a tool for exactly that.) The last line shows the real power: we *computed* a new value from an existing one. If `age` changes, we edit one place and everything downstream updates.

Two naming rules to remember (there are more, but these carry you far):

- Names are written in `snake_case`: lowercase words joined by underscores, e.g. `birth_year`, `coffee_price`.
- Choose names that say what the value *means* — `temperature` beats `t`.

## 3. Data types — not all values are alike

Look at a spreadsheet from any field: some columns hold whole numbers (participants), some hold decimals (measurements), some hold text (city names), and some hold yes/no answers. Python makes the same distinction. The four types you will meet every single day:

| Type | Python name | Example | Typical use |
|------|-------------|---------|-------------|
| Whole number | `int` | `42` | counts, years, IDs |
| Decimal number | `float` | `19.99` | measurements, prices, averages |
| Text ("string") | `str` | `"Berlin"` | names, labels, answers |
| Yes/No | `bool` | `True`, `False` | flags, comparison results |

You can always *ask* Python what type a value has, using `type()`:

````python
participants = 42          # int
average_temp = 21.5        # float
city = "Berlin"            # str
survey_complete = True     # bool

print(type(participants))
print(type(average_temp))
print(type(city))
print(type(survey_complete))
````

**Reading the output:** `<class 'int'>` looks more dramatic than it is. The word in quotes is the answer — `int`, `float`, `str`, `bool`. (`class` is simply Python's own word for "kind of value"; you can ignore it for now.)

**Why care about types?** Because types decide what you can *do* with a value. You can multiply two numbers, but `"Berlin" * "Hamburg"` makes no sense — and Python will say so. The most common beginner trip-up: `"42"` (with quotes) is *text*, not a number. `"42" + 1` produces an error; `42 + 1` produces `43`.

## 4. Arithmetic — Python as a very reliable calculator

Python knows the operators from school maths:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | addition | `7 + 3` | `10` |
| `-` | subtraction | `7 - 3` | `4` |
| `*` | multiplication | `7 * 3` | `21` |
| `/` | division | `7 / 2` | `3.5` |
| `**` | power | `2 ** 10` | `1024` |

Let's combine variables and arithmetic for a tiny real calculation — what a daily coffee habit costs per year:

````python
coffee_price = 3.20        # euros per cup
cups_per_day = 2
days_per_year = 365

yearly_cost = coffee_price * cups_per_day * days_per_year
print("Coffee per year, in euros:")
print(yearly_cost)
````

**Reading the output:** `2336.0` — that daily habit is a small holiday per year! But why the `.0` on a euro amount? Because `coffee_price` is a `float`, and as soon as a decimal number is involved, the result is a decimal number too. (Division with `/` *always* gives a `float`, even for `6 / 2`.) The types from Section 3 quietly at work.

## 5. Strings and f-strings — talking about your results

A bare number (`2336.0`) is not a result — *"Coffee costs you €2,336.00 per year"* is. Combining text and values is such a common job that Python has a dedicated tool for it: the **f-string**.

Put an `f` before the opening quote, and you can embed any variable — or a small calculation — inside curly braces `{ }`. After a colon you can add a **format instruction**; the two you will use most in data work are `:.2f` (exactly two decimal places) and `:,` (thousands separator).

````python
name = "Alice"
age = 25

# Embedding values — even small calculations work inside { }
print(f"{name} is {age} years old.")
print(f"Next year, {name} will be {age + 1}.")

# Formatting numbers
print(f"Coffee per year: {yearly_cost:.2f} euros")      # exactly 2 decimals
print(f"Nicely readable: €{yearly_cost:,.2f}")          # separators + 2 decimals
````
