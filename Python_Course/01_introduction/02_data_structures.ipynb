## 1. Why data structures? One value per box does not scale

Suppose you collect a feedback score (1–5) from every participant in this room. With Notebook 1's tools you would write:

```python
score_1 = 4
score_2 = 5
score_3 = 3      # ... and now imagine 300 participants
```

Three things are already broken: you cannot average 300 hand-named variables, you cannot add a 301st participant without editing your code, and nothing in the code says these values belong together.

Real data arrives in **collections** — a column of measurements, a row describing one person, a table of both. Python gives you three everyday collections, and this table is the map for the rest of the notebook:

|  | Looks like | You look things up by | Can it change? | Use it for |
|---|---|---|---|---|
| **List** | `[4, 5, 3]` | **position**: `scores[0]` | yes — items change, list grows | a *series* of similar things: measurements, answers, rows |
| **Tuple** | `(52.52, 13.40)` | **position**: `location[0]` | no — fixed once created | a small *fixed record*: coordinates, a pair, a function's several results |
| **Dictionary** | `{"age": 31}` | **name (key)**: `person["age"]` | yes | *one thing* described by labelled fields |

One sentence to carry away: **lists number their contents, dictionaries name them, tuples refuse to be edited.** Come back to this table whenever you are unsure which one you want.



## 2. Lists — many values in numbered compartments

A **list** is one box with numbered compartments. You build it with square brackets and commas:

```
scores  =  [  4  ,  5  ,  3  ,  5  ,  4  ]
position:     0     1     2     3     4
```

A compartment's position number is called its **index**, and you use it to pull one value out: `scores[2]`. One convention trips up absolutely everybody at first: **Python starts counting at 0.** The first item is item `0`, the second is item `1`. So the *fifth* item of our list is `scores[4]`.

````python
scores = [4, 5, 3, 5, 4]          # feedback scores from five participants

print(f"the whole list   : {scores}")
print(f"first item [0]   : {scores[0]}")      # counting starts at 0!
print(f"third item [2]   : {scores[2]}")
print(f"last item [-1]   : {scores[-1]}")     # a negative index counts from the END
print(f"how many items   : {len(scores)}")      # len() counts the items
print(f"average score    : {sum(scores) / len(scores):.2f}")   # sum() adds them all up
````


**Reading the output:** `scores[0]` gives `4`, the *first* score, and `scores[2]` gives `3`, the *third*. The negative index is a lovely shortcut: `scores[-1]` means "the last one" without your having to know how long the list is (`scores[-2]` is the second-to-last). The last two lines use two ready-made tools that Python hands you for free: **`len()`** counts the items (`5` — which is also the number of survey responses) and **`sum()`** adds up every number in the list.

Put them together and you get the real payoff: `sum(scores) / len(scores)` is the average — `4.20` — and that single line works unchanged for 5 scores or 5,000. That is what a collection buys you: **one name for the whole dataset.**

````python
# 1) Change one item — participant 3 corrected their score afterwards:
scores[2] = 4
print(f"after the correction : {scores}")

# 2) Grow the list — a late participant hands in their form:
scores.append(5)
print(f"after .append(5)     : {scores}")
print(f"number of scores now : {len(scores)}")

# 3) Slicing, as a one-line taster — the first three scores:
print(f"the first three      : {scores[0:3]}")
````


**Reading the output:** the third item really did change from `3` to `4`, and after `.append(5)` the list is one item longer — `len()` now reports `6`. Lists are **mutable**: the box itself is modified, so every variable pointing at that list sees the new contents. (A live-session warning: run this cell twice and `.append()` runs twice, so the list keeps growing — exactly the *run order matters* point from Notebook 1.)

Line 3 is your first **slice**: `scores[0:3]` means "from position 0 up to *but not including* position 3" — three items. That asymmetry has a logic to it, and Track 2's [`03_lists_data_structures.ipynb`](../02_advanced_self_learning/03_lists_data_structures.ipynb) gives you the full model. For today, read `scores[0:3]` as "the first three" and move on.

## 3. Tuples — a small record that must not change

A list is a shopping list: things get added, crossed off, rewritten. A **tuple** is more like a printed label or a stamped coordinate — created once, then fixed. You build one with **round** brackets:

```python
location = (52.52, 13.40)     # (latitude, longitude) of Berlin
```

You read items exactly as in a list (`location[0]`), and `len()` works too. What you *cannot* do is change an item. That sounds like a limitation; it is actually the point. Some things belong together and must never drift apart — the two halves of a coordinate, the three parts of a date — and a tuple guarantees that nobody edits one half by accident.

````python
location = (52.52, 13.40)          # round brackets → a tuple

print(f"the whole tuple : {location}")
print(f"latitude  [0]   : {location[0]}")
print(f"longitude [1]   : {location[1]}")
print(f"how many parts  : {len(location)}")

# Unpacking: hand each part its own variable name, in one line
lat, lon = location
print(f"unpacked        : lat = {lat}, lon = {lon}")

# A tuple refuses to be edited. Uncomment the next line, run it, and read the error:
# location[0] = 48.14
````


**Reading the output:** indexing and `len()` behave just like a list. (Python prints `13.4`, not `13.40` — numbers don't remember how you typed them, exactly as in Notebook 1.) The interesting line is `lat, lon = location` — **unpacking**. Python matches the variables on the left to the items on the right, in order, so you get two clearly named values instead of having to remember that latitude was "the one at position 0". Unpacking is the reason tuples feel pleasant in practice.

If you uncomment the last line, Python stops you:

```text
TypeError: 'tuple' object does not support item assignment
```

Read it as *"you may not assign into a tuple"*. That is the meaning of **immutable**. So when is a tuple the right choice?

- **Fixed records** — a coordinate `(lat, lon)`, an RGB colour `(31, 119, 180)`, a date `(2026, 8, 22)`.
- **Values that belong together** and should never be edited separately.
- **Returning several results from a function** — `return x, y` hands back a tuple; you will meet it in Track 2's [`05_functions_and_modules.ipynb`](../02_advanced_self_learning/05_functions_and_modules.ipynb).

Rule of thumb: a **list** is a series of *many similar* things; a **tuple** is *one thing with a fixed number of parts*.

## 4. Dictionaries — look things up by name, not by number

Look again at `participant[1]`. Was position 1 the age or the field of study? With a list or tuple *you* have to remember the meaning of every position — and in real projects, you won't.

A **dictionary** fixes that by labelling each value. You write it with **curly** braces as `key: value` pairs, and you look things up by the label — the **key** — instead of a number:

```
        list / tuple                        dictionary
   position 0 → "Amina"              "name"  → "Amina"
   position 1 → 31                   "age"   → 31
   position 2 → "Sociology"          "field" → "Sociology"
```

Think of a paper form, or an index card: each field has a printed name, and you read the box next to the name you want. The order of the fields does not matter; their names do.

````python
participant = {
    "name": "Amina",
    "age": 31,
    "field": "Sociology",
    "attended_before": False,
}

print(f"the whole record : {participant}")
print(f"one value by key : {participant['age']}")
print(f"how many fields  : {len(participant)}")
print(f"{participant['name']} is {participant['age']} and works in {participant['field']}.")
````


**Reading the output:** `participant["age"]` reads almost like English — "the age of this participant" — and, unlike `participant[1]`, it will still make sense to you next month. Note that each field keeps its own type: `"Amina"` is a `str`, `31` an `int`, `False` a `bool`. A dictionary happily mixes types, because its job is to describe *one thing* from several angles. `len()` on a dictionary counts the **fields**, not the characters.

(One small syntax detail: inside an f-string written with double quotes, the key needs *single* quotes — `f"{participant['age']}"`. Outside f-strings, `participant["age"]` with double quotes is the usual style.)

Records are rarely finished when you first write them. Let's add a field, correct one, and ask what the record contains:

````python
# Add a NEW field — just assign to a key that isn't there yet:
participant["country"] = "Kenya"

# Update an EXISTING field — exactly the same syntax:
participant["age"] = 32

print(f"the record now : {participant}")
print(f"the labels     : {participant.keys()}")
print(f"the contents   : {participant.values()}")

# Membership: does this record even have such a field?
print(f'"country" in participant -> {"country" in participant}')
print(f'"email"   in participant -> {"email" in participant}')

# participant["email"]    # ← uncomment, run, and read the KeyError
````


**Reading the output:** adding and updating look identical — `participant["country"] = "Kenya"` created a new field because no such key existed, while `participant["age"] = 32` overwrote the old value. `.keys()` gives you the field names and `.values()` their contents; Python shows them wrapped as `dict_keys([...])` and `dict_values([...])`, which you can simply read as "the labels" and "the contents" of the record.

`in` answers a question you will ask constantly with real data: *is this field present at all?* `"country" in participant` is `True`, `"email" in participant` is `False`. It matters, because reading a key that isn't there does not give you an empty value — it stops your program:

```text
KeyError: 'email'
```

That `KeyError` is one of the most common errors in real data work, because real records have missing fields. Which is exactly why `in` earns its keep: ask *before* you read, and your code never crashes on a record that is merely incomplete.

## 5. The bridge — a list of dictionaries is a table

Now combine two of the three. One dictionary describes one participant; a **list of dictionaries** describes *all* of them — one dictionary per row.

````python
participants = [
    {"name": "Amina", "age": 32, "field": "Sociology"},
    {"name": "Ben",   "age": 27, "field": "Law"},
    {"name": "Chen",  "age": 41, "field": "Medicine"},
]

print(f"rows (participants) : {len(participants)}")
print(f"the second row      : {participants[1]}")
print(f"one single cell     : {participants[1]['name']}")   # row 1, column "name"
````


Read that output as a table: the **list** holds the rows in order, each **dictionary** is one row, and the keys (`name`, `age`, `field`) are the column names — which is why `participants[1]["name"]` picks out exactly one cell. This is not a toy analogy: a list of dictionaries is the shape data actually arrives in from surveys, databases and web APIs.

And it is precisely the shape **pandas** takes over in Notebook 5. There, the same rows become a `DataFrame` that you can inspect, filter, average per column and plot — without ever writing a position lookup by hand again.

## 🧠 Key takeaways

1. **One value per variable does not scale.** Real data comes in collections, and Python's three everyday ones are the list, the tuple and the dictionary.
2. **List** — `[4, 5, 3]`: a series of similar things, looked up by **position** (`scores[0]`, `scores[-1]` for the last), measured with `len()`, summed with `sum()`, changed item-by-item, grown with `.append()`. `scores[0:3]` slices out the first three.
3. **Tuple** — `(52.52, 13.40)`: a small fixed record. **Immutable** — assigning into it raises `TypeError` — and best used with **unpacking**: `lat, lon = location`.
4. **Dictionary** — `{"age": 31}`: one thing with labelled fields, looked up by **key** (`person["age"]`). The same syntax adds and updates; `.keys()` / `.values()` show labels and contents. A missing key raises `KeyError`, so test with `in` before you read.
5. **Lists number their contents, dictionaries name them, tuples refuse to be edited.** Pick by the job, not by habit.
6. A **list of dictionaries is a table** — one dict per row, keys as column names. Exactly what pandas turns into a `DataFrame` in Notebook 5.
