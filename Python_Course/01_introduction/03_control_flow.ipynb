## 1. Asking yes/no questions — comparisons and booleans

Before Python can *decide* anything, it must *ask a question* whose answer is `True` or `False`. Those two values are the **booleans** you met in Notebook 1, and every decision in every program on Earth boils down to them.

| Operator | Question it asks   | Example    | Answer  |
|----------|--------------------|------------|---------|
| `==`     | equal?             | `5 == 5`   | `True`  |
| `!=`     | not equal?         | `5 != 3`   | `True`  |
| `<`      | less than?         | `3 < 5`    | `True`  |
| `>`      | greater than?      | `3 > 5`    | `False` |
| `<=`     | less or equal?     | `5 <= 5`   | `True`  |
| `>=`     | greater or equal?  | `4 >= 6`   | `False` |

> ⚠️ **The classic beginner mix-up:** one `=` **assigns** a value (`x = 5` means "store 5 in x"). Two `==` **compares** (`x == 5` asks "is x equal to 5?"). Python will refuse to run `if x = 5:` — a helpful safety net.

You can also **combine** questions with `and`, `or`, `not` — they work just like in English.

## 2. `if / elif / else` — teaching Python to decide

### The intuition: a decision tree

Think of how a doctor triages patients, or how you decide what to wear: you walk down a list of questions **in order** and take the **first branch that applies** — then you stop asking.

```text
                 Is it below 5°C? ──── yes ──→  "winter coat"
                        │ no
                 Is it below 15°C? ─── yes ──→  "light jacket"
                        │ no
                        └────────────────────→  "t-shirt"
```

That is *exactly* what `if / elif / else` does: `if` asks the first question, each `elif` ("else-if") asks the next one, and `else` catches everything that fell through.

### Indentation — how Python knows what belongs to a decision

Notebook 1 promised that the spaces at the start of a line are meaningful. Here is why: Python does **not** use brackets to group code — it uses **indentation**.

```text
if temperature < 5:
····print("Winter coat!")      ← indented 4 spaces → belongs to the if
····print("And a scarf.")      ← same indent → also belongs to the if
print("Have a nice day.")     ← back at the left edge → ALWAYS runs
```

Three rules prevent 90% of beginner errors:

1. A line that opens a block **ends with a colon `:`** (`if ...:`, `else:`, and soon `for ...:` and `while ...:`).
2. Everything inside the block is indented by **4 spaces** (the notebook does this for you when you press Enter after a `:`).
3. The block ends when the indentation goes **back to the left**.

See `IndentationError` or `SyntaxError: expected ':'`? Check exactly these three things.


````python
temperature = 12   # try changing this to 3, then to 25, and re-run!

if temperature < 5:
    print("🧥 Winter coat!")
elif temperature < 15:
    print("🧊 Light jacket weather.")
else:
    print("👕 T-shirt is fine.")

print(f"(Decision made for {temperature}°C — this line always runs.)")
````

**One branch, and only one.** Python checks the conditions from top to bottom and runs the **first** one that is `True` — the rest are skipped entirely. That's why the order matters: at 3°C, *both* `temperature < 5` and `temperature < 15` are true, but only the first branch fires. So order your questions from **strictest to loosest**.

## 3. `for` loops — doing something for *every* item

### The intuition: a conveyor belt

Your list is a conveyor belt. A `for` loop picks up each item, **one at a time**, hands it to you under a name you choose, and runs your indented block once per item.

```text
for city in cities:        ← "for each item (call it `city`) in the list..."
····print(city)            ← ...run this block once per item
```

Same indentation rules as `if`: colon at the end, 4 spaces inside.

And when there is no list — you simply want to repeat something *n* times — use **`range(n)`**: it generates the numbers `0, 1, 2, …, n-1` for you (computers love starting at 0!).


````python
cities = ["Berlin", "Nairobi", "Osaka", "Lima"]

for city in cities:
    print(f"Hello, {city}!")

# range(n): repeat n times, counting 0 ... n-1
print()
for i in range(5):
    print(f"Round {i}")

# range(start, stop) — starts at `start`, stops JUST BEFORE `stop`
print()
for year in range(2024, 2028):
    print(f"Year {year}")
````

### The accumulator pattern — building up a result step by step

This is *the* most important loop pattern in data science. To sum a list by hand you would keep a **running total** on a piece of paper and add each number to it. Code does exactly the same:

1. **Before** the loop: create a variable that starts "empty" (e.g. `total = 0`).
2. **Inside** the loop: update it with each item.
3. **After** the loop: the result is ready.

Step 2 is written so often that Python has a shorthand for it: `total += steps` means exactly `total = total + steps`. You will see `+=` in every loop from here on.


````python
daily_steps = [8500, 12300, 6200, 9800, 15000]

total = 0                    # 1. start empty, BEFORE the loop
for steps in daily_steps:
    total = total + steps    # 2. update once per item
    # shortcut for the same thing:  total += steps

print(f"Total steps this week : {total}")            # 3. use the result
print(f"Average per day       : {total / len(daily_steps):.0f}")   # :.0f → no decimals
````

## 4. `while` loops — repeating until something changes

A `for` loop needs to know **what to walk through**: a list, or `range(n)`. Sometimes you don't know how many rounds it will take — you only know **when to stop**.

| | You know… | Use |
|---|---|---|
| `for` | the collection (or the number of rounds) | `for x in my_list:` |
| `while` | only the **stopping condition** | `while still_short:` |

Read `while` as *"as long as this is true, keep going"*. Python checks the condition, runs the block, checks again, runs again… and stops the moment the condition becomes `False`.

> ⚠️ **The runaway loop.** Something inside the block **must** move you towards the stopping condition. Forget it, and the loop never ends:
>
> ```python
> balance = 0
> while balance < 1000:
>     print("saving...")     # balance never changes → this runs forever!
> ```
>
> It is harmless: press the **⏹ stop button** next to the running cell (in Colab: *Runtime → Interrupt execution*). Then fix the line that should have changed the condition. Before running a `while`, ask yourself once: *"which line brings me closer to stopping?"*


````python
# How many months of saving until we reach 1000 €?
# We do NOT know the number of rounds in advance — that is exactly what `while` is for.

target = 1000
balance = 0
months = 0

while balance < target:      # "as long as we are still short of the target..."
    balance += 150           # save 150 € this month  ← this line moves us towards stopping
    months += 1              # ...and count the month

print(f"Reached {balance} € after {months} months (target was {target} €).")
````

## 5. Putting it together — a mini weather report 🌤️

Time for the payoff: a `for` loop, an `if / elif / else` **inside** it, an accumulator, and f-strings — the whole notebook in ~15 lines. Note the **double indentation**: the `if` is indented once (inside the loop), its branches twice (inside the `if`, inside the loop).

One new trick first: we have **two lists that belong together pair by pair** — `days[0]` ("Mon") goes with `temps[0]` (8°C), and so on. To visit the *pairs*, we loop over the **positions** with `range(len(days))` and use each position `i` to look up `days[i]` *and* `temps[i]` — Notebook 2's list indexing doing double duty.

Read the code and predict what it will print — *then* run it.


````python
days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temps = [8, 19, 24, 31, 22, 28, 33]

print("=== Weekly Weather Report ===")

hot_days = 0                               # accumulator: starts at 0 BEFORE the loop
for i in range(len(days)):                 # i = 0, 1, 2, ... 6 — one position per day
    temp = temps[i]
    if temp < 10:
        category = "cold 🧊"
    elif temp <= 25:
        category = "pleasant 🙂"
    else:
        category = "hot 🔥"
        hot_days += 1                      # label it AND count it
    print(f"  {days[i]}: {temp:>2}°C — {category}")   # :>2 right-aligns, so the °C line up

print("=============================")
print(f"Hot days this week: {hot_days} of {len(days)}")
````

**Take a second to appreciate this.** One loop + one decision + one counter — and you have a program that works just as well on 7 days or 7 *million* days of data. That scalability is *the* reason we automate analysis with code instead of reading spreadsheets by eye.

## 🧠 Key takeaways

1. **Booleans** (`True`/`False`) answer yes/no questions built with `==`, `!=`, `<`, `>` — and `and` / `or` / `not` combine them. Remember: `=` assigns, `==` compares.
2. **`if / elif / else`** is a decision tree: Python runs the **first true branch** and skips the rest — so order your conditions from strictest to loosest.
3. **Indentation is the grammar of Python**: a colon `:` opens a block, 4 spaces mark what's inside, and going back to the left edge closes it.
4. **`for` loops** visit every item of a list once; **`range(n)`** counts `0 … n-1` when there's no list.
5. **`while` loops** repeat as long as a condition holds — for when you know the *stopping condition* but not the number of rounds. Always make sure something inside the loop moves towards it.
6. The **accumulator pattern** (initialise before the loop → update with `+=` inside → use after) builds sums and counts — combine it with `if` to filter.
