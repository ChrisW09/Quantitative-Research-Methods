## 1. The problem functions solve — copy-paste pain

Let's feel the pain first; the cure will make more sense afterwards.

You are given three weather-station readings in **degrees Fahrenheit** and asked to report them in **degrees Celsius**. With only what you know so far, you would write the conversion once per station — which in practice means writing it once and pasting it twice:


````python
# Three readings in °F, converted to °C — the same arithmetic, copy-pasted.
station_a_c = 68 * 5/9
station_b_c = 95 * 5/9
station_c_c = 41 * 5/9

print(f"Station A: {station_a_c:.1f} °C")
print(f"Station B: {station_b_c:.1f} °C")
print(f"Station C: {station_c_c:.1f} °C")
````

Three numbers, printed without complaint. Except they are **all wrong**: the Fahrenheit scale does not start at zero, so you must subtract 32 **before** multiplying by 5/9. Python had no way of knowing that — it did exactly the arithmetic we asked for.

So the requirement changes (as it always does) and the formula must be fixed. Note what that means with copy-pasted code: **you must fix it in every copy**. Below, we patch stations A and B, get interrupted by a colleague, and never come back to station C:


````python
station_a_c = (68 - 32) * 5/9    # patched ✅
station_b_c = (95 - 32) * 5/9    # patched ✅
station_c_c = 41 * 5/9           # ...never got patched ❌

print(f"Station A: {station_a_c:.1f} °C")
print(f"Station B: {station_b_c:.1f} °C")
print(f"Station C: {station_c_c:.1f} °C")

print()   # an empty print() just leaves a blank line
print(f"What station C should say: {(41 - 32) * 5/9:.1f} °C")
````

Look at what just happened. Station C reports **22.8 °C** when the true value is **5.0 °C** — a 17.8-degree error in a number that would go straight into a report. No error message, no warning: the code ran perfectly and lied to us, because two lines now follow the new rule and one still follows the old one.

That is not a story about being careless. It is a story about **arithmetic that lives in three places**. With three copies you might catch it; with thirty you will not.

### The cure: write the recipe once, give it a name

A **function** is a named recipe. Think of a machine on a workbench:

```text
        argument                                   returned value
   68  ──────────►  ┌───────────────────────┐  ──────────────────►  20.0
   (input slot)     │  to_celsius(f):       │      (output tray)
                    │    (f - 32) * 5/9     │
                    └───────────────────────┘
```

You build the machine **once**. After that you drop values into the input slot and collect results from the output tray — and if the recipe ever needs fixing, there is exactly **one** place to fix it.

Here is the anatomy, then the real thing:

```text
def to_celsius(fahrenheit):          ← `def`, a name, parameters in (), colon
····return (fahrenheit - 32) * 5/9   ← the body: indented 4 spaces, `return` hands a value back
```

The indentation is Notebook 3's rule again: the colon opens a block, and the indented lines are what belongs inside it.


````python
def to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9


# Defining the function printed nothing — we only built the machine.
# Now we USE it, once per station, with a different argument each time:
print(f"Station A: {to_celsius(68):.1f} °C")
print(f"Station B: {to_celsius(95):.1f} °C")
print(f"Station C: {to_celsius(41):.1f} °C")
````

**20.0, 35.0, 5.0** — correct, all three, including station C. And notice the important part: the formula now appears **exactly once** in this notebook. Station C can no longer disagree with station A, because they are no longer separate copies; they are the same recipe used three times.

The vocabulary, so we can talk about it precisely:

| Term | In our example | Meaning |
|---|---|---|
| `def` | `def to_celsius(...)` | "define a function" — the keyword that starts a recipe |
| **name** | `to_celsius` | how you call it later; name it after *what it produces* |
| **parameter** | `fahrenheit` | the blank on the recipe card, filled in at call time |
| **body** | the indented line | the work, indented 4 spaces under the colon |
| **argument** | the `68` in `to_celsius(68)` | the actual value you hand in |
| **`return`** | `return (fahrenheit - 32) * 5/9` | the value the function hands back |


## 2. Two parameters — and what `return` really buys you

A recipe can have more than one blank. List the parameters inside the brackets, separated by commas, and hand in the arguments in the **same order**.

Here is a function every data analysis needs sooner or later: what percentage is this part of that whole?

````python
def share_percent(part, whole):
    return part / whole * 100


# Two survey groups: 18 of 25 agreed, and 31 of 40 agreed.
print(f"Group A: {share_percent(18, 25):.1f} % agreed")
print(f"Group B: {share_percent(31, 40):.1f} % agreed")

# Because the function RETURNS a value, that value is just a number —
# so we can keep computing with it:
gap = share_percent(31, 40) - share_percent(18, 25)
print(f"Group B leads Group A by {gap:.1f} percentage points")
````

That last line is the whole point of `return` — and mixing up `return` and `print` is *the* most common beginner bug, so let's make it explicit.

## 3. ⚠️ `return` vs `print` — the one distinction to get right

- **`print`** puts characters on your screen. That is *all* it does. The value is gone the moment you have read it.
- **`return`** hands the value back **to your code**, so you can store it in a variable, add it to something, or feed it into another function.

A function that only prints looks fine — you *see* a number, after all — but it secretly returns nothing. Python spells "nothing" as `None`, and `None` cannot be added, averaged, or plotted.

````python
def show_celsius(fahrenheit):
    print((fahrenheit - 32) * 5/9)      # only DISPLAYS the number


def get_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9      # HANDS THE NUMBER BACK


shown = show_celsius(68)     # the 20.0 below is printed by the function itself
print(f"shown = {shown}")    # → None: the function handed nothing back
print(type(shown))           # type() from Notebook 1 confirms it

got = get_celsius(68)
print(f"got   = {got}")
print(f"got + 5 = {got + 5}")   # a real number, usable in further arithmetic

# Uncomment the next line, run it, and read the error — then comment it out again:
# print(shown + 5)   # TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'
````

Both functions *displayed* `20.0`, but only one of them **gave** us `20.0`. Rule of thumb: a function that computes something should `return` it; print at the *end*, where you present results — not in the middle, where you are still calculating.

## 4. A whole list as input — and defaults for convenience

Nothing says an argument has to be a single number. Hand a function a **list** and put a `for` loop in the body, and you have named a whole summarising step: `average(readings)` says what it does, and the loop mechanics disappear inside.

### A parameter can have a default value

Write `rate=0.1` in the parameter list and that parameter becomes **optional**: callers who omit it get the default, callers who care can override it. Same recipe, less typing for the common case.


````python
def average(readings):
    total = 0                      # the accumulator pattern from Notebook 3,
    for value in readings:         # now safely tucked inside a named function
        total = total + value
    return total / len(readings)


week = [18.5, 21.0, 19.5, 25.0, 23.5, 17.0, 20.5]   # °C, one reading per day
print(f"Weekly average: {average(week):.2f} °C")

# The returned value is an ordinary number, so it works anywhere a number does:
if average(week) > 20:
    print("A mild week overall.")


# A default value makes the second argument optional:
def tip(amount, rate=0.1):
    return amount * rate


print(f"Tip on 50 € at the default 10 %: {tip(50):.2f} €")
print(f"Tip on 50 € at a generous 20 %:  {tip(50, 0.2):.2f} €")
````

`average(week)` reads like a sentence, and `tip(50)` needs no second thought. Both are ordinary functions — nothing new, just applied to a list and given a default.

## 5. Why functions structure code

Removing duplication was only the beginning. The deeper gain is that **naming a step turns a wall of code into a description of your analysis**. `average(week)`, `share_percent(18, 25)`, `classify(reading)` — a reader can follow the argument without decoding a single line of arithmetic. That buys three things at once:

- **Readable.** The name says the intent; the mechanics stay out of the way until you need them.
- **Testable.** You can check `to_celsius(212)` and expect exactly `100.0`. A formula smeared across forty lines has nothing you can point at and check.
- **Reusable.** Written once, correct once. The station-C bug from Section 1 is not merely fixed — it has become *impossible*.

And this is not a beginner's habit you will outgrow. It is how the whole ecosystem is built: every `pd.read_csv(...)` and `.describe()` you will write in Notebook 5 is **someone else's function** — a named recipe, with parameters, that hands a value back. Learning to write your own is learning to read theirs.

## 🧠 Key takeaways

1. **Copy-pasted logic rots.** When a rule changes you must fix every copy, and the copy you miss produces wrong numbers *silently* — no error, just a report that disagrees with itself.
2. **A function is a named recipe**: `def name(parameters):` followed by an indented body. Defining it does nothing visible; **calling** it does the work.
3. **Parameters are the blanks; arguments are what you put in them** — one, two, or a whole list.
4. **`return` hands a value back to your code; `print` only shows it.** A function that just prints returns `None`, and `None` cannot be added, averaged, or plotted. Compute with `return`, present with `print`.
5. **Default values** (`def tip(amount, rate=0.1)`) make a parameter optional — convenience for the common case, flexibility when you need it.
6. **Naming a step makes an analysis readable, testable, and reusable** — and every library call you will ever make is somebody else's function.
