## 2. Variables — naming pieces of data

A **variable** is a name we attach to a value so we can reuse it later. Think of variables as *labelled boxes*:

```
┌──────────┐    ┌───────┐    ┌──────────┐
│  "Alice" │    │   25  │    │   5.6    │
└────┬─────┘    └───┬───┘    └─────┬────┘
     │              │              │
   name            age           height
```

In Python you create a variable by **assigning** a value with `=`:

````python
name = "Alice"      # a string of text
age = 25            # an integer (whole number)
height = 1.70       # a float (decimal number)
is_student = True   # a boolean (True/False)

# Display all four neatly using f-strings (more on these later)
print(f"Name      : {name}")
print(f"Age       : {age}")
print(f"Height (m): {height}")
print(f"Student?  : {is_student}")
````

### Naming rules (and good style)

Python is strict about what is a *valid* name and the community has strong opinions about what is a *good* name:

| Rule                                                   | Example                |
|--------------------------------------------------------|------------------------|
| Must start with a letter or `_`                        | `age`, `_temp`         |
| Can contain letters, digits, `_`                       | `score_1`, `user_id`   |
| Case sensitive                                         | `Age` ≠ `age`          |
| Cannot be a reserved word (`for`, `class`, `if`, …)    | ❌ `class = 5`         |
| Use `snake_case` for variables and functions (PEP 8)   | ✅ `customer_count`    |

> 💡 **Choose names that describe meaning, not type.**  Prefer `temperature_celsius` over `t` or `temp_float`.

## 3. Core data types

In data science you'll meet four basic types over and over again:

| Type     | Python name | Example          | Used for                                 |
|----------|-------------|------------------|------------------------------------------|
| Integer  | `int`       | `42`             | counts, indices, IDs                      |
| Float    | `float`     | `19.99`          | measurements, probabilities, prices       |
| String   | `str`       | `"hello"`        | text, category labels, column names       |
| Boolean  | `bool`      | `True`, `False`  | flags, filters, comparison results        |

There is also a special placeholder, `None`, meaning "no value yet" — you'll see it for missing data later in the course.

````python
# Use type() to ask Python: "what kind of value is this?"
count        = 42
price        = 19.99
city         = "Berlin"
has_discount = True
nothing      = None

for value in (count, price, city, has_discount, nothing):
    print(f"{repr(value):>10}  ->  {type(value).__name__}")
````

**Reading the output.** Each row shows the value (in `repr()` form so the quotes around strings are visible) and the Python type name. Notice how `42` and `42.0` are *different* types — they look similar but behave subtly differently when you divide them.

### Converting between types

You can explicitly convert a value with `int()`, `float()`, `str()`, `bool()`. This is called *casting*:

````python
# From string to number — common when reading CSV files
age_text = "25"
age_num  = int(age_text)
print(f"'25' as text  ->  {age_text!r:>6}  (type {type(age_text).__name__})")
print(f"25  as int    ->  {age_num:>6}  (type {type(age_num).__name__})")

# The string must "look like" the target type:
# int("3.9") would raise a ValueError — go via float() first:
print()
print(f"float('3.9')      = {float('3.9')}")
print(f"int(float('3.9')) = {int(float('3.9'))}")

# From float to int — note the loss of precision
print()
print(f"int(3.9)   = {int(3.9)}    <- truncates toward zero, does NOT round!")
print(f"int(-3.9)  = {int(-3.9)}   <- toward zero for negatives too")
print(f"round(3.9) = {round(3.9)}   <- this is what you usually want")

# round() on exact halves: Python rounds to the nearest EVEN number
print()
print(f"round(0.5) = {round(0.5)},  round(1.5) = {round(1.5)},  round(2.5) = {round(2.5)}")

# bool() in Python: 0, 0.0, '', None, [] are 'falsy'; almost everything else is 'truthy'
print()
print(f"bool(0)       = {bool(0)}")
print(f"bool('')      = {bool('')}")
print(f"bool('False') = {bool('False')}  <- non-empty string, so truthy!")
````

> ⚠️ **Four conversion pitfalls worth memorising.**
> 1. `int(3.9)` returns `3`, not `4` — casting *truncates toward zero* (so `int(-3.9)` is `-3`). Use `round()` when you actually want rounding.
> 2. `round()` has its own surprise: exact halves go to the nearest **even** number — `round(2.5)` is `2`, not `3`. This is *banker's rounding*: over many numbers it avoids a systematic upward bias (imagine a bank that always rounded interest up in its own favour…).
> 3. `int("3.9")` raises a `ValueError` — the string must look like an integer. Convert via `float()` first.
> 4. `bool("False")` is `True` — any *non-empty* string is truthy, regardless of what it says. When a CSV file hands you `"False"` as text, compare it (`value == "True"`) instead of casting it.

## 4. Arithmetic operations

Python supports the operators you would expect, plus two that are particularly useful in data science:

| Operator | Meaning             | Example         | Result |
|----------|---------------------|-----------------|--------|
| `+`      | addition            | `7 + 3`         | `10`   |
| `-`      | subtraction         | `7 - 3`         | `4`    |
| `*`      | multiplication      | `7 * 3`         | `21`   |
| `/`      | division (float!)   | `7 / 3`         | `2.333…` |
| `//`     | floor division      | `7 // 3`        | `2`    |
| `%`      | modulo (remainder)  | `7 % 3`         | `1`    |
| `**`     | exponent (power)    | `2 ** 10`       | `1024` |

`%` and `//` come up surprisingly often — for example to test whether a number is even (`n % 2 == 0`).

````python
a, b = 17, 5

print(f"a + b  = {a + b}")
print(f"a - b  = {a - b}")
print(f"a * b  = {a * b}")
print(f"a / b  = {a / b:.4f}    <-- regular division always returns a float")
print(f"a // b = {a // b}        <-- floor division: drops the remainder")
print(f"a % b  = {a % b}        <-- remainder of a / b")
print(f"a ** b = {a ** b}    <-- a to the power b")
````

### Operator precedence — just like in maths

`*` and `/` bind tighter than `+` and `-`. Use parentheses whenever there is any doubt — it costs nothing and prevents bugs.

````python
# These two are NOT the same!
print("2 + 3 * 4   =", 2 + 3 * 4)     # 14
print("(2 + 3) * 4 =", (2 + 3) * 4)   # 20

# In data work, parentheses make the *intent* explicit:
revenue = 1000
tax_rate = 0.19
net = revenue / (1 + tax_rate)
print(f"Net revenue: €{net:,.2f}")
````

## 5. Strings — working with text

Strings are how we hold text: names, labels, file paths, category values. Data is rarely "just numbers": a column of cities, a customer name, a status flag — they all show up as strings.

````python
first_name = "Ada"
last_name  = "Lovelace"

# 1. Concatenation with +
full_name = first_name + " " + last_name
print(f"Full name : {full_name}")

# 2. Length
print(f"Letters    : {len(full_name)}")

# 3. Case methods (return NEW strings — they don't modify the original)
print(f"Upper      : {full_name.upper()}")
print(f"Lower      : {full_name.lower()}")
print(f"Title case : {full_name.title()}")

# 4. Stripping whitespace — VERY common when cleaning CSV data
messy = "   data science   "
print(f"Stripped   : {messy.strip()!r}")

# 5. Splitting and joining — also common in data cleaning
csv_row = "Ada,Lovelace,1815"
parts = csv_row.split(",")
print(f"Split      : {parts}")
````

> 💡 String methods return a *new* string. Strings in Python are **immutable** — `name.upper()` does not change `name`; it gives you back a new value that you must assign if you want to keep it.

## 6. F-strings — formatting output

F-strings (Python 3.6+) are the modern way to build strings that include variable values. They start with `f` and use `{ ... }` to embed expressions.

The general pattern is:

```
f"... {expression:format_spec} ..."
```

The `:format_spec` part is optional but very useful for clean output.

````python
student = "Alice"
score   = 87.6543
pi      = 3.14159265

# Basic embedding
print(f"Hello, {student}!")

# Two-decimal places
print(f"Score : {score:.2f}")

# As a percentage (multiplies by 100 and adds %)
print(f"Pass rate : {0.873:.1%}")

# Width and alignment
print(f"|{'left':<10}|{'center':^10}|{'right':>10}|")
print(f"|{student:<10}|{score:^10.1f}|{pi:>10.4f}|")

# Thousands separator (great for money / large counts)
revenue = 1_234_567.89
print(f"Revenue : €{revenue:,.2f}")

# Debug form (Python 3.8+): show variable name + value
print(f"{score=:.2f}")
````

**Cheat-sheet of the most useful format specs**

| Spec       | Meaning                              | Example output |
|------------|--------------------------------------|----------------|
| `:.2f`     | fixed-point, 2 decimals              | `3.14`         |
| `:.1%`     | percent, 1 decimal                   | `12.3%`        |
| `:,`       | thousands separator                  | `1,234,567`    |
| `:>10`     | right-align in width 10              | `        42`   |
| `:<10`     | left-align in width 10               | `42        `   |
| `:^10`     | centre-align in width 10             | `    42    `   |
| `:.3e`     | scientific notation, 3 decimals      | `1.235e+03`    |

## 7. Updating variables

Variables can be reassigned. A very common shortcut is the *augmented assignment* operator `+=` (and friends):

````python
balance = 1000
print(f"Start          : €{balance:,.2f}")

# Long form
balance = balance + 150
print(f"After +150     : €{balance:,.2f}")

# Augmented form — same thing, less typing
balance += 200   #  equivalent to: balance = balance + 200
print(f"After +200 more: €{balance:,.2f}")

# Works for the other operators too
counter = 1
counter *= 2   # counter = counter * 2
counter *= 2
counter *= 2
print(f"Counter doubled 3 times: {counter}")
````

## 8. A first applied example — investment portfolio

Let's combine everything: variables, arithmetic, strings and f-strings to compute the value and one-year dividend yield of a small portfolio. This is the *kind* of small calculation a financial analyst might do every day.

````python
# === A tiny portfolio of three stocks ===
# (price per share, shares held, annual dividend per share)
stock_a_price, stock_a_shares, stock_a_dividend = 150.75, 50, 2.30
stock_b_price, stock_b_shares, stock_b_dividend = 89.20, 75, 1.45
stock_c_price, stock_c_shares, stock_c_dividend = 234.50, 25, 4.10

# Position value = price * shares
value_a = stock_a_price * stock_a_shares
value_b = stock_b_price * stock_b_shares
value_c = stock_c_price * stock_c_shares
total_value = value_a + value_b + value_c

# Dividend income = dividend per share * shares
div_a = stock_a_dividend * stock_a_shares
div_b = stock_b_dividend * stock_b_shares
div_c = stock_c_dividend * stock_c_shares
total_div = div_a + div_b + div_c

# Yield = annual dividends / total value
yield_pct = total_div / total_value

# === Pretty report ===
print("=" * 50)
print("📊 PORTFOLIO REPORT".center(50))
print("=" * 50)
print(f"{'Position':<10}{'Value':>15}{'Dividend':>15}")
print("-" * 50)
print(f"{'Stock A':<10}{value_a:>15,.2f}{div_a:>15,.2f}")
print(f"{'Stock B':<10}{value_b:>15,.2f}{div_b:>15,.2f}")
print(f"{'Stock C':<10}{value_c:>15,.2f}{div_c:>15,.2f}")
print("-" * 50)
print(f"{'TOTAL':<10}{total_value:>15,.2f}{total_div:>15,.2f}")
print(f"\nDividend yield: {yield_pct:.2%}")
````

**What just happened?**

1. We stored 9 raw numbers as variables with clear names.
2. We *derived* new variables (`value_a`, `div_a`, …) using arithmetic — no copy-paste of numbers.
3. We assembled a nicely aligned report using f-string width and decimal specs.

Notice how every number on the report comes from a single calculation. If a stock price changes you only need to edit it in one place — that's the power of variables.

## 9. Common beginner pitfalls

Spotting these will save you hours of debugging later.

1. **Type vs. text.** `"25"` is a string, `25` is an integer. `"25" + 5` raises `TypeError`. Convert with `int("25")` first.
2. **Integer vs. float division.** `5 / 2` is `2.5`. `5 // 2` is `2`. With negatives the two rounding directions disagree: `//` *floors* (`-5 // 2` is `-3`) while `int()` *truncates* (`int(-5 / 2)` is `-2`). Choose deliberately.
3. **Strings are immutable.** `name.upper()` returns a new string. Always assign it back if you want to keep the change: `name = name.upper()`.
4. **Re-using variable names.** Assigning `list = [...]` overwrites the built-in `list` function. Avoid names that shadow built-ins (`list`, `dict`, `str`, `type`, `sum`, `id`, `input`).
5. **Comparing floats with `==`.** Because of binary floating-point, `0.1 + 0.2 == 0.3` is `False`. Use `abs(a - b) < 1e-9` for "close enough" comparisons.

Run the next cell to *see* a couple of these in action:

````python
# Pitfall 2: division
print("5 / 2  =", 5 / 2)      # 2.5  (float)
print("5 // 2 =", 5 // 2)     # 2    (int — the remainder is dropped)

# ...and with NEGATIVE numbers, // and int() disagree:
print("-5 // 2     =", -5 // 2)       # -3  (floors: toward minus infinity)
print("int(-5 / 2) =", int(-5 / 2))   # -2  (truncates: toward zero)

# Pitfall 5: floating-point surprises
a, b = 0.1 + 0.2, 0.3
print(f"\n0.1 + 0.2 = {a!r}")
print(f"is it == 0.3?  {a == b}")
print(f"close enough?  {abs(a - b) < 1e-9}")

# Pitfall 1: TypeError on text + number
try:
    "25" + 5
except TypeError as e:
    print(f"\nTypeError caught: {e}")
````

**Reading the output — why does `0.1 + 0.2` come out as `0.30000000000000004`?**

Computers store numbers in *binary*, and `0.1` is a repeating fraction in binary — just like `1/3 = 0.3333…` never terminates in decimal. The computer has to cut the digits off somewhere, and that tiny leftover error becomes visible when such numbers are added. This is not a Python bug; every mainstream language behaves the same way. Two practical consequences:

- never compare floats with `==` — use a tolerance, as in `abs(a - b) < 1e-9`;
- financial software often sidesteps the problem by counting in whole cents (integers) or using Python's `decimal` module.

The negative-division lines deserve a second look too: `-5 // 2` gives `-3` (floor division always rounds *down*, toward minus infinity) while `int(-5 / 2)` gives `-2` (casting always cuts *toward zero*). For positive numbers the two happen to agree — which is exactly why this bug hides until negative data shows up.

## 🧠 Key takeaways

1. **Variables** are labelled containers — pick descriptive `snake_case` names.
2. The four core types you'll meet daily are `int`, `float`, `str`, `bool`. `None` represents "no value".
3. **Convert** between types with `int()`, `float()`, `str()`, `bool()`; `int()` truncates toward zero, and `round()` sends exact halves to the nearest *even* number (banker's rounding).
4. **Arithmetic** in Python supports `+ - * / // % **`. `/` always returns a float; `//` floors — it rounds *down* even for negatives (`-5 // 2` is `-3`).
5. **F-strings** are the standard way to format output. Master `.2f`, `,`, `%`, alignment specs — you will use them constantly.
6. Strings are **immutable** — methods return new strings. Always assign the result back if you want to keep it.
7. Watch out for **floating-point comparison** issues: prefer `abs(a - b) < eps` over `a == b` for floats.
