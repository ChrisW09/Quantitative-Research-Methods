## 1. What is a dictionary?

Think of a dictionary as a **labelled filing cabinet**: each drawer has a **label** (the *key*) and contains some *value*.

```
       ┌───── student ─────┐
"name"   →  "Alice"
"age"    →  20
"major"  →  "Computer Science"
"gpa"    →  3.8
```

- Keys must be **unique** and **immutable** (strings, numbers, tuples — *not* lists).
- Values can be anything (numbers, strings, lists, even other dictionaries).
- Dictionaries are **ordered** since Python 3.7 (keys keep their insertion order).

````python
# Creating a dictionary
student = {
    "name":  "Alice",
    "age":   20,
    "major": "Computer Science",
    "gpa":   3.8,
}

print(student)
print(f"type = {type(student).__name__}, size = {len(student)}")

# Empty dict
inventory = {}
print(f"empty: {inventory}")
````

## 2. Accessing values

Use `dict[key]` to look up a value. Like list-indexing, but with a meaningful label instead of a position.

````python
student = {"name": "Alice", "age": 20, "major": "CS", "gpa": 3.8}

print(f"Name : {student['name']}")
print(f"Age  : {student['age']}")
print(f"GPA  : {student['gpa']}")

# What if the key doesn't exist?
try:
    print(student["height"])
except KeyError as e:
    print(f"\nKeyError: missing key {e}")

# Safer: .get() returns None (or a default) when the key is missing
print(f"\n.get('height') -> {student.get('height')}")
print(f".get('height', 'unknown') -> {student.get('height', 'unknown')}")

# Or simply ask first — `in` checks the KEYS of a dictionary
print(f"\n'gpa' in student    -> {'gpa' in student}")
print(f"'height' in student -> {'height' in student}")
````

> 💡 **Rule of thumb:** use `d[key]` when a missing key would be a **bug you want to hear about** — the loud `KeyError` points you straight at it. Use `d.get(key, default)` when missing keys are **normal** (optional fields, messy real-world data). And `key in d` lets you ask before you touch.

## 3. Adding, updating and removing entries

````python
person = {"name": "Bob", "age": 25}
print(f"Original : {person}")

# Add new keys
person["city"] = "Berlin"
person["job"]  = "Data Analyst"
print(f"After add: {person}")

# Update an existing key (same syntax — there's no difference)
person["age"] = 26
print(f"After update: {person}")

# Update multiple keys at once
person.update({"city": "Munich", "salary": 65_000})
print(f"After bulk update: {person}")

# Remove keys
removed = person.pop("salary")   # remove AND return the value
print(f"Popped 'salary' = {removed}")
print(f"After pop: {person}")

del person["city"]
print(f"After del 'city': {person}")
````

## 4. Useful methods: `keys()`, `values()`, `items()`

````python
inventory = {"apples": 50, "bananas": 30, "oranges": 25, "grapes": 40}

print(f"Keys   : {list(inventory.keys())}")
print(f"Values : {list(inventory.values())}")
print(f"Items  : {list(inventory.items())}")

# Numbers you can compute directly
print(f"\nTotal stock : {sum(inventory.values())}")
print(f"Most stocked: {max(inventory, key=inventory.get)}")   # neat trick
````

> 🎯 **`max(d, key=d.get)`** finds the key whose value is largest — a one-liner version of "which item has the most stock?". Two things are happening: `max` walks a dictionary's **keys** (as we'll see in §5), and `key=` takes a *function* that turns each key into the thing to compare on. Note there are **no parentheses** after `inventory.get` — we hand `max` the function itself and let it do the calling, once per key. Passing functions as arguments is a habit worth getting used to; §8 does it again with `lambda`.

## 5. Iterating over a dictionary

Three flavours, depending on what you need:

````python
grades = {"Math": 85, "Science": 92, "English": 78, "History": 88}

# 1) Iterating directly gives KEYS
print("Keys only:")
for subject in grades:
    print(f"  {subject}")

# 2) Just the values
print("\nValues only:")
for score in grades.values():
    print(f"  {score}")

# 3) Both at once — the most common pattern
print("\nKey + value (using .items()):")
for subject, score in grades.items():
    print(f"  {subject:<10} {score}")
````

## 6. Dictionary comprehensions

The same `{expression for ... in ... if ...}` syntax as list comprehensions, but produces a dictionary.

````python
# Build a dict from scratch
squares = {x: x**2 for x in range(1, 6)}
print("Squares:", squares)

# Transform values
celsius     = {"morning": 20, "noon": 30, "evening": 25}
fahrenheit  = {k: (v * 9/5 + 32) for k, v in celsius.items()}
print(f"\nC: {celsius}")
print(f"F: {fahrenheit}")

# Filter — keep only items meeting a condition
scores  = {"Alice": 85, "Bob": 92, "Charlie": 67, "Diana": 78, "Eve": 95}
high    = {name: s for name, s in scores.items() if s >= 80}
print(f"\nHigh scorers: {high}")

# Invert — swap keys and values (works when values are unique)
inverted = {v: k for k, v in scores.items()}
print(f"\nInverted: {inverted}")
````

## 7. Nested dictionaries — real-world structured data

In practice, a dictionary value is often *another* dictionary. This is how you'd represent records from a database, JSON from an API or settings from a config file.

````python
students_db = {
    "S001": {
        "name":   "Alice Johnson",
        "age":    20,
        "major":  "Computer Science",
        "grades": {"Math": 85, "Science": 92, "English": 78},
    },
    "S002": {
        "name":   "Bob Smith",
        "age":    22,
        "major":  "Mathematics",
        "grades": {"Math": 95, "Science": 88, "English": 82},
    },
    "S003": {
        "name":   "Carol Lee",
        "age":    21,
        "major":  "Physics",
        "grades": {"Math": 90, "Science": 95, "English": 75},
    },
}

# Drill down with chained keys
print(students_db["S001"]["grades"]["Math"])     # → 85

# Walk the database and compute each student's average grade
for sid, info in students_db.items():
    g = info["grades"]
    avg = sum(g.values()) / len(g)
    print(f"  [{sid}] {info['name']:<14} avg = {avg:5.2f}")
````

### 🧭 How not to get lost in nested data

A chain like `blob["results"][0]["readings"]["temp_c"]` is how you *finish*, never how you *start*. Guess wrong on an unfamiliar structure and Python tells you almost nothing useful: a `KeyError` names one missing key without saying which level was wrong, and a `TypeError: list indices must be integers` just means you treated a list as a dict.

The reliable method is to **peel one level at a time**, asking two questions at every stop:

1. **What type am I holding?** `type(x).__name__` — a `dict` (look things up by key) or a `list` (index by position)?
2. **What's inside it?** For a dict: `list(x.keys())`. For a list: `len(x)`, then look at `x[0]` **only** — in real data one element tells you the shape of all of them.

Then descend into the single key you care about and repeat. Four stops are enough to map the blob below. Watch the third city as you go: it has no `humidity`. Real records are **ragged**, which is precisely the situation §2's `.get()` was made for — and the moment where the `[]`-vs-`.get()` decision becomes real rather than theoretical.

````python
# Pretend this blob just arrived from a web API and nobody told you its shape.
api_blob = {
    "status": "ok",
    "meta": {"page": 1, "per_page": 3},
    "results": [
        {"id": 101, "city": "Berlin",   "readings": {"temp_c": 21.4, "humidity": 63}},
        {"id": 102, "city": "Lisbon",   "readings": {"temp_c": 27.9, "humidity": 48}},
        {"id": 103, "city": "Helsinki", "readings": {"temp_c": 14.2}},  # humidity missing!
    ],
}

# STEP 1 — what am I even holding?
print("1)", type(api_blob).__name__, "with keys:", list(api_blob.keys()))

# STEP 2 — descend ONE level into the interesting key, then ask again
results = api_blob["results"]
print("2)", type(results).__name__, "of length", len(results))

# STEP 3 — it's a list: inspect ONE element, never the whole thing
first = results[0]
print("3)", type(first).__name__, "with keys:", list(first.keys()))

# STEP 4 — keep peeling until you reach plain values
print("4)", type(first["readings"]).__name__, "with keys:", list(first["readings"].keys()))

# Only NOW is chaining safe, because every level has been verified:
print("\nLisbon temp:", api_blob["results"][1]["readings"]["temp_c"])

# And here is the decision from §2, made on real data:
print("\nAll cities:")
for r in results:
    temp = r["readings"]["temp_c"]                # REQUIRED field -> [] : a miss is a bug, crash loudly
    hum = r["readings"].get("humidity", "n/a")    # OPTIONAL field -> .get() : a miss is expected
    print(f"  {r['city']:<9} {temp:>5.1f} °C   humidity: {hum}")
````

## 8. List of dictionaries — the shape of (almost) all data

Here comes the single most important idea in this notebook. When you fetch data from a web API or read a CSV file, what you get is almost always a **list of dictionaries**: one dict per **row**, with the column names as keys.

```
employees = [
    {"name": "Alice", "department": "Engineering", "salary": 75000, "years": 3},   # ← row 1
    {"name": "Bob",   "department": "Marketing",   "salary": 65000, "years": 5},   # ← row 2
    ...
]
```

Three names, one shape:

| You call it…        | It looks like…                            |
|---------------------|-------------------------------------------|
| a **table**         | rows × columns, like a spreadsheet        |
| a **list of dicts** | `[{...}, {...}, ...]` in Python           |
| **JSON**            | what practically every web API sends you  |

Learn to manipulate this one shape and you can handle real data *today* — and in Notebook 7 ([`07_pandas_essentials.ipynb`](07_pandas_essentials.ipynb)) you'll see that `pd.DataFrame(employees)` turns it into a pandas table in a single call.

````python
employees = [
    {"name": "Alice",   "department": "Engineering", "salary": 75000, "years": 3},
    {"name": "Bob",     "department": "Marketing",   "salary": 65000, "years": 5},
    {"name": "Charlie", "department": "Engineering", "salary": 80000, "years": 4},
    {"name": "Diana",   "department": "Sales",       "salary": 60000, "years": 2},
    {"name": "Eve",     "department": "Engineering", "salary": 90000, "years": 7},
    {"name": "Frank",   "department": "Marketing",   "salary": 70000, "years": 6},
]

# Average salary
avg_salary = sum(e["salary"] for e in employees) / len(employees)
print(f"Average salary : €{avg_salary:,.0f}")

# Highest-paid employee
top = max(employees, key=lambda e: e["salary"])
print(f"Top earner     : {top['name']} (€{top['salary']:,})")

# Sort by years of experience, descending
ranked = sorted(employees, key=lambda e: e["years"], reverse=True)
print("\nRanked by experience:")
for r in ranked:
    print(f"  {r['name']:<8} {r['years']:>2}y in {r['department']}")
````

> 💡 **What is `lambda e: e["salary"]`?** A *lambda* is a tiny unnamed function written inline: "given a row `e`, hand back `e['salary']`". Passing it as `key=` tells `max` / `sorted` **which field to compare on** — Python calls it once per row and ranks the rows by what comes back. We'll study lambdas properly in Notebook 5 ([`05_functions_and_modules.ipynb`](05_functions_and_modules.ipynb)); until then, `key=lambda row: row["column"]` is a recipe worth memorising.

````python
# The claim was "a list of dicts IS a table". Let's read the same data both ways.
rows = employees          # one dict per row

# --- COLUMN view: pull one field out of every row ---
print("column 'salary':", [e["salary"] for e in rows])
print("column 'name'  :", [e["name"] for e in rows])

# --- The column NAMES are simply the keys of a row ---
columns = list(rows[0].keys())
print("\ncolumns        :", columns)

# --- ROW view: print the whole thing as a table, header straight from those keys ---
header = " | ".join(f"{c:<11}" for c in columns)
print("\n" + header)
print("-" * len(header))
for r in rows:
    print(" | ".join(f"{str(r[c]):<11}" for c in columns))
````

> 💡 **Read that output twice — nothing was converted.** No new data structure was built; the list of dicts *already was* the table, and we only chose which direction to read it in.
>
> - The **header row** is `rows[0].keys()` — a record's keys *are* the column names.
> - A **column** is one comprehension across all rows: `[e["salary"] for e in rows]`.
> - A **row** is one dict — the unit you filter, print, or hand back to an API.
>
> This is the whole trick behind pandas. `pd.DataFrame(rows)` runs essentially the loop you just read (Notebook 7, [`07_pandas_essentials.ipynb`](07_pandas_essentials.ipynb)), and `df["salary"]` is that comprehension with a nicer face on it. `df.iloc[0]` gives you back the row; `df["salary"]` gives you back the column — same duality, same data.
>
> One caveat that matters in practice: the aligned print above uses `r[c]` and so assumes **every row carries every key**. Ragged records (§7) would raise `KeyError` — swap in `r.get(c, "—")` and the table still prints with a gap where the value is missing. That is exactly what pandas does when it puts `NaN` in a cell.

## 9. Counting and grouping — dictionaries' superpower

Counting categories and grouping rows by a column are textbook dictionary jobs. The pattern below appears in **every** data-cleaning project.

````python
# Count survey responses
responses = ["Yes", "No", "Yes", "Maybe", "Yes", "No", "Yes", "Maybe", "Yes", "No"]

counts = {}
for r in responses:
    counts[r] = counts.get(r, 0) + 1     # idiom: get-or-default 0, then add 1

print(f"Counts: {counts}")
````

````python
# Even cleaner with collections.Counter (built into Python)
from collections import Counter

responses = ["Yes", "No", "Yes", "Maybe", "Yes", "No", "Yes", "Maybe", "Yes", "No"]
print(Counter(responses))                # like a dict, with a few extras
print(Counter(responses).most_common(2)) # top-2 most common values
````

> 💡 **Compare the two cells: same answer, one loop fewer.** That is the win — and it's worth having written the manual version first, because a `Counter` is not magic, it is precisely that four-line loop wrapped up.
>
> What `Counter` adds: it **is** a dictionary (a `dict` subclass), so `counts["Yes"]`, `.items()`, `.values()` and dict comprehensions all still work — plus a missing key returns `0` instead of raising `KeyError` (no `.get()` dance), and `.most_common(n)` hands you the top *n* already sorted.
>
> Where it stops: `Counter` **counts occurrences of things**. It cannot sum a numeric column by group (that's Exercise 4.1 — back to the `get`-or-default idiom), and it cannot collect the rows themselves. For that you need the grouping pattern next.

````python
# Group employees by department — list of dicts → dict of lists
employees = [
    {"name": "Alice",   "department": "Engineering", "salary": 75000},
    {"name": "Bob",     "department": "Marketing",   "salary": 65000},
    {"name": "Charlie", "department": "Engineering", "salary": 80000},
    {"name": "Diana",   "department": "Sales",       "salary": 60000},
    {"name": "Eve",     "department": "Engineering", "salary": 90000},
    {"name": "Frank",   "department": "Marketing",   "salary": 70000},
]

# The manual pattern — same spirit as counting, but collecting into lists
by_dept = {}
for e in employees:
    dept = e["department"]
    if dept not in by_dept:      # first time we see this department?
        by_dept[dept] = []       # ...start an empty list for it
    by_dept[dept].append(e["name"])

print("Employees by department:")
for dept, names in by_dept.items():
    print(f"  {dept:<12}: {names}")

# defaultdict removes the "is this key new?" dance: missing keys start as an empty list
from collections import defaultdict

salaries_by_dept = defaultdict(list)
for e in employees:
    salaries_by_dept[e["department"]].append(e["salary"])

print("\nAverage salary by department:")
for dept, salaries in salaries_by_dept.items():
    print(f"  {dept:<12}: €{sum(salaries)/len(salaries):,.0f}")
````

> 🔭 **Where this is heading.** Everything in this section is *one line* in pandas:
>
> ```python
> pd.DataFrame(employees).groupby("department")["salary"].mean()
> ```
>
> That's Notebook 7 ([`07_pandas_essentials.ipynb`](07_pandas_essentials.ipynb)). Doing it by hand first is not wasted effort: you now know exactly what `groupby` does under the hood — most pandas confusion comes from skipping this step.

## 10. Lists vs dictionaries — when to use which?

| Use a **list** when…                                       | Use a **dict** when…                                      |
|------------------------------------------------------------|-----------------------------------------------------------|
| Order matters (time series, ranks, instructions).          | Order is incidental; lookup-by-name matters.              |
| You'll access items by **position** (`x[0]`, `x[1]`).      | You'll access items by **label** (`x["name"]`).            |
| Duplicates are fine and meaningful.                        | Keys must be unique.                                       |
| Items are *similar* and interchangeable.                   | Items are *named* with different meanings.                 |

> 🎯 In data science, *records* (rows) are usually dicts; *columns* are usually lists/arrays.

## 11. JSON — dictionaries on the wire

**JSON** is the universal data-exchange format of the web, and it is essentially Python dictionaries and lists written out as text. Four functions cover almost everything:

| Call | Direction | Use it for |
|---|---|---|
| `json.dumps(obj)` | Python → JSON **string** | sending over a network, logging, embedding in a notebook |
| `json.loads(text)` | JSON **string** → Python | anything that arrived as text |
| `json.dump(obj, f)` | Python → **file** | `with open("out.json", "w") as f: json.dump(data, f)` |
| `json.load(f)` | **file** → Python | `with open("data.json") as f: data = json.load(f)` |

The **`s` stands for *string***: `dumps`/`loads` work on text in memory, `dump`/`load` on an open file. We stay with the string versions below — this notebook runs in Colab, where there are no local files worth reading, and the file versions are the same functions with a file handle instead of a string.

This is not a side trip. It is how real data reaches you:

```python
import requests
data = requests.get("https://api.example.com/employees").json()   # → a list of dicts
```

`.json()` there is just `json.loads` applied to the response body, and `data` comes out as exactly the shape you practised on in §8: a list of dicts you can filter, sort, group — or hand straight to `pd.DataFrame`.

````python
import json

# A single record is a dict...
api_response = {
    "city": "Berlin",
    "temperature_c": 21.4,
    "conditions": ["cloudy", "windy"],
    "forecast": {"tomorrow": "rain", "weekend": "sunny"},
}
print(json.dumps(api_response, indent=2))

# Round-trip: dict → JSON string → dict
s = json.dumps(api_response)
back = json.loads(s)
print(f"\nRound-trip equal? {back == api_response}")

# ...and a TABLE is a list of dicts — this is literally what most APIs send you
team = [
    {"name": "Alice", "department": "Engineering", "salary": 75000},
    {"name": "Bob",   "department": "Marketing",   "salary": 65000},
]
print("\n" + json.dumps(team, indent=2))

# But JSON has fewer types than Python, so a round trip is not always lossless:
tricky = {"tags": ("a", "b"), 1: "an int key", "nothing": None, "flag": True}
print("\nbefore round-trip:", tricky)
print("after  round-trip:", json.loads(json.dumps(tricky)))
````

> ⚠️ **JSON is *nearly* Python, not exactly.** The first round trip printed `True`, but that is a property of *that* object, not a promise. Look at the last two lines of output: the tuple came back as a **list** and the integer key `1` came back as the **string** `"1"`.
>
> | You put in | You get back | Why |
> |---|---|---|
> | `("a", "b")` tuple | `["a", "b"]` list | JSON has arrays; it has no tuples |
> | `1` used as a **key** | `"1"` | JSON object keys are always strings |
> | `None`, `True` | `None`, `True` (written `null`, `true` in the text) | just a different spelling — these survive |
> | `set`, `datetime`, NumPy array | **`TypeError`** | no JSON equivalent; convert to a list or string first |
>
> The practical rules: after loading JSON, don't assume a type you haven't checked (the §7 peeling method is how you check). And when you *write* JSON, hand it only dicts, lists, strings, numbers, booleans and `None`.

The error message is `TypeError: 'str' object is not callable` — read it literally: something tried to **call** a string. That something is `sorted`, which calls whatever you pass as `key` once per item. So there are two issues:

1. `key="price"` is invalid — `key` expects a *function* that maps an item to the value to compare on, not the name of a field. Use `lambda p: p["price"]`.

top3 = sorted(products, key=lambda p: p["price"], reverse=True)[:3]

## 🧠 Key takeaways

1. Dictionaries store **key → value** pairs; keys must be unique and immutable.
2. Use `d[k]` when a missing key means a **bug you want to hear about**, and `d.get(k, default)` when a missing key is **expected** — that decision, not the syntax, is the skill.
3. `.keys()`, `.values()`, `.items()` are how you iterate; **`.items()` is by far the most useful**.
4. A **list of dictionaries** *is* a table: the keys of a row are the column names, and a column is a comprehension across rows.
5. Map unfamiliar nested data by **peeling one level at a time** — `type(x)`, then `.keys()` for a dict or `len(x)` and `x[0]` for a list — instead of guessing a chain of keys.
6. The `dict.get(k, 0) + value` idiom is the classic "count / sum by group" pattern; `Counter` and `defaultdict` are that same loop, prepackaged.
7. Dictionaries map directly to **JSON** — the shape every web API sends — but JSON has fewer types, so tuples become lists and keys become strings on the round trip.
