## 2. The Figure / Axes mental model

Matplotlib has two layers you need to understand:

```
┌────────────── Figure (the whole window) ──────────────┐
│                                                        │
│   ┌────── Axes (one plot inside) ──────┐               │
│   │                                    │               │
│   │   data lives here                  │               │
│   │                                    │               │
│   └────────────────────────────────────┘               │
│                                                        │
│   ┌────── Axes (another plot) ─────────┐               │
│   │                                    │               │
│   └────────────────────────────────────┘               │
└────────────────────────────────────────────────────────┘
```

- A **Figure** is the canvas — the sheet of paper. It holds one or many plot areas.
- An **Axes** is *one* plot area — a coordinate system with an x-axis, a y-axis and data on it.
  (Confusing name: an *Axes* is a whole plot, not a single axis.)

`fig, ax = plt.subplots()` hands you both in one line: the canvas, and the single plot area on it.

### Which object owns which method?

Once the two objects are clear, you almost never have to guess where a method lives:

| You want to change…                       | Ask the…   | Examples |
|-------------------------------------------|------------|----------|
| anything *inside* one plot area           | **Axes**   | `ax.plot`, `ax.bar`, `ax.set_xlabel`, `ax.set_ylim`, `ax.legend`, `ax.annotate` |
| something spanning the *whole canvas*     | **Figure** | `fig.suptitle`, `fig.colorbar`, `fig.savefig`, `fig.tight_layout` |

Rule of thumb: on an Axes, `ax.set_<thing>(...)` sets it and `ax.get_<thing>()` reads it back.

### The one aside: the state-based style

Matplotlib also has an older, "invisible object" style, where `plt` quietly keeps a *current
figure* and every call lands on it:

```python
# State-based (pyplot) style — acceptable for a three-line throwaway:
plt.plot([1, 2, 3], [2, 4, 8])
plt.title("quick look")
plt.show()
```

It reads shorter, and you will meet it in every old tutorial — which is why it appears here
once. But "the current figure" is guesswork the moment you have two figures or a 2×2 grid,
and the bug in Exercise 9.5 is exactly the kind of bug it invites.

**From here on, every chart in this notebook is built with `fig, ax = plt.subplots()` and drawn
with `ax.…`.** The only pyplot calls that survive are `plt.tight_layout()` and `plt.show()`,
which act on the figure you have just built (`fig.tight_layout()` is the explicit equivalent).

## 3. Your first line plot

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, color=BAS_BLUE, linewidth=2)
ax.set_title("First plot: sin(x)")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
plt.tight_layout()
plt.show()
````

Those few lines are essentially every plot you will ever make:

1. `fig, ax = plt.subplots(...)`  — create a Figure and one Axes.
2. `ax.plot(...)`  — draw the data.
3. `ax.set_title / set_xlabel / set_ylabel`  — label it.
4. `plt.tight_layout()`  — fix overlapping labels.
5. `plt.show()`  — render.

Everything else is variations on this recipe.

### 🏷️ Labelling discipline — three rules worth making a habit

1. **Every axis gets a label**, with units where they exist — "Revenue (€M)", not just "Revenue".
2. **The title states the takeaway** whenever there is one: *"Sales doubled in 2024"* beats *"Sales over time"*. A descriptive title is the fallback; a missing title is a bug.
3. **Add a legend the moment a plot has more than one data series.**

A chart that follows these rules can be read without you standing next to it to explain it — which is the whole point of making one.

## 4. Multiple lines, legends, line styles

````python
x  = np.linspace(0, 2 * np.pi, 200)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(2 * x)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, y1, label="sin(x)",    color=BAS_BLUE, linewidth=2)
ax.plot(x, y2, label="cos(x)",    color=BAS_AMBER, linewidth=2, linestyle="--")
ax.plot(x, y3, label="sin(2x)",   color=BAS_GREEN, linewidth=2, linestyle=":")
ax.axhline(0, color=BAS_GREY, linewidth=0.7)

ax.set_title("Three trigonometric functions")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend(loc="upper right")
plt.tight_layout()
plt.show()
````

> 💡 Common line styles: `"-"` solid, `"--"` dashed, `":"` dotted, `"-."` dash-dot. Markers: `"o"` circle, `"s"` square, `"^"` triangle, `"x"` cross, `"."` point.

For colour, use named CSS colours (`"crimson"`), hex codes (`"#416bcc"`) or any standard matplotlib name. Stick to a small, consistent palette — this course uses the `BAS_*` constants defined in the setup cell — and your audience will thank you.

## 5. 🧭 Which chart answers which question?

Before asking *"how do I draw this?"*, ask *"what question should the reader be able to answer?"* — the question determines the chart:

| Your question | Chart type | matplotlib call |
|---|---|---|
| How does something change **over time**? (trend) | line | `ax.plot` |
| Are two numeric variables **related**? | scatter | `ax.scatter` |
| How do **categories compare**? | bar / horizontal bar | `ax.bar` / `ax.barh` |
| How is one variable **distributed**? | histogram, box plot | `ax.hist` / `ax.boxplot` |
| What **share** does each part contribute? (composition) | stacked bar — or a pie for 2–4 slices only | `ax.bar(..., bottom=…)` / `ax.pie` |
| What does a whole **matrix** of values look like? | heatmap | `ax.imshow` |

Two rules that come with the table:

- **One question per chart.** If your chart answers two, it usually answers neither well — split it.
- **Composition is the treacherous row.** People compare lengths far better than angles, so a
  pie chart stops working past a handful of slices; a sorted horizontal bar chart of the same
  shares almost always reads better. You will test this yourself in Exercise 9.3.

If you cannot say which question your chart answers, you are not ready to plot yet.

The next sections walk through this table row by row — for each chart, pay as much attention to the *question it answers* as to the syntax.

## 6. Scatter plots — for relationships between two variables

````python
rng = np.random.default_rng(42)
n = 80

# Synthetic relationship: y is roughly proportional to x plus noise
x = rng.uniform(0, 10, size=n)
y = 2.0 * x + rng.normal(0, 2.5, size=n)

# Add a third "feature" used for colour
quality = rng.integers(1, 6, size=n)

fig, ax = plt.subplots(figsize=(8, 5))
sc = ax.scatter(x, y, c=quality, cmap="Blues", s=60, edgecolor=BAS_INK, alpha=0.8)

ax.set_title("Scatter: y ≈ 2x + noise (colour = quality 1–5)")
ax.set_xlabel("x")
ax.set_ylabel("y")

# Best-fit line (a one-line linear regression with numpy)
coef = np.polyfit(x, y, deg=1)
xs   = np.linspace(x.min(), x.max(), 50)
ax.plot(xs, coef[0] * xs + coef[1], color=BAS_AMBER, linestyle="--",
        label=f"fit: y = {coef[0]:.2f}x + {coef[1]:.2f}")

fig.colorbar(sc, ax=ax, label="Quality rating")
ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
````

**Reading the chart.** The cloud of points slopes upward → x and y are positively correlated. The colour-encoded third variable shows whether there is any visible pattern in *quality* (here: not really). The amber dashed line is the fitted regression. We will meet `np.polyfit` again in Notebook 13 inside scikit-learn's `LinearRegression`.

## 7. Bar charts — for comparing categories

````python
cities      = ["Berlin", "Munich", "Hamburg", "Cologne", "Frankfurt"]
populations = [3.66, 1.49, 1.85, 1.09, 0.76]   # in millions

# Cities have no natural order, so sort by value: the reader gets the ranking
# for free instead of having to hunt for it.
order       = np.argsort(populations)[::-1]        # descending
cities      = [cities[i] for i in order]
populations = [populations[i] for i in order]

fig, ax = plt.subplots(figsize=(8, 4.5))
bars = ax.bar(cities, populations,
              color=[BAS_BLUE, BAS_AMBER, BAS_GREEN, BAS_RED, BAS_PURPLE],
              edgecolor=BAS_INK)

# Annotate each bar with its value
for bar, val in zip(bars, populations):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.05,
            f"{val:.2f}", ha="center", fontsize=10)

ax.set_title("Berlin has almost twice the population of any other German city")
ax.set_ylabel("Population (millions)")
ax.set_ylim(0, max(populations) * 1.15)          # bars must start at zero
plt.tight_layout()
plt.show()
````

**Reading a bar chart.** Bar *length* encodes the quantity, so the value axis has to start at
zero — a bar chart with a cut-off axis is not a style choice, it is a false statement (we come
back to this in section 13). Two habits are on display above:

- **Sorted by value.** "Berlin, Munich, Hamburg, …" is not an order any reader cares about.
  Sorting turns "hunt for the biggest" into "look at the left". Keep the original order only when
  the categories *have* one — months, age bands, a survey scale from *strongly disagree* upwards.
- **A title that says something.** "Population of German cities" would only repeat the axes.
  The title above states the finding, so the chart still makes its point when someone drops it
  into a slide deck without you.

### Horizontal bar charts

Use `barh` whenever the category labels are long — they read better horizontally.

````python
skills = ["Python", "SQL", "Statistics", "Machine Learning",
          "Communication", "Domain Knowledge", "Visualisation"]
importance = [95, 88, 92, 85, 90, 78, 82]

fig, ax = plt.subplots(figsize=(8, 5))
order = np.argsort(importance)
ax.barh(np.array(skills)[order], np.array(importance)[order],
        color=BAS_BLUE, edgecolor=BAS_INK)

ax.set_title("Self-reported importance of data-science skills (0–100)")
ax.set_xlabel("Importance")
ax.set_xlim(0, 100)
plt.tight_layout()
plt.show()
````

## 8. Histograms — for distributions

````python
rng = np.random.default_rng(0)

# Two simulated test-score populations
normal_class = rng.normal(loc=75, scale=10, size=400)
honors_class = rng.normal(loc=88, scale=5,  size=200)

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(normal_class, bins=25, alpha=0.65, label="Normal class",
        color=BAS_BLUE, edgecolor=BAS_INK)
ax.hist(honors_class, bins=25, alpha=0.65, label="Honors class",
        color=BAS_AMBER, edgecolor=BAS_INK)

# Vertical lines for means
ax.axvline(normal_class.mean(), color=BAS_BLUE, linestyle="--",
           label=f"Normal mean = {normal_class.mean():.1f}")
ax.axvline(honors_class.mean(), color=BAS_AMBER, linestyle="--",
           label=f"Honors mean = {honors_class.mean():.1f}")

ax.set_title("Honors students score ~13 points higher — with half the spread")
ax.set_xlabel("Score (0–100)")
ax.set_ylabel("Number of students")
ax.legend()
plt.tight_layout()
plt.show()
````

**Reading a histogram.** The x-axis is the value, the y-axis the count of observations in each bin. Two overlapping histograms with `alpha=0.65` is a clean way to compare two distributions. Vertical dashed lines highlight the means.

> 🎯 **The bin count is a choice, not a fact.** Re-run the cell above with `bins=5` and then
> `bins=200`: too few bins hide the two peaks, too many turn the distribution into noise. If a
> feature in your data only looks bimodal at one specific bin count, it probably isn't.

## 9. Box plots — distribution at a glance

````python
rng = np.random.default_rng(1)
classes = ["Math", "Science", "English", "History"]
scores  = [rng.normal(75, 12, 60),
           rng.normal(80,  8, 60),
           rng.normal(72, 15, 60),
           rng.normal(78, 10, 60)]

fig, ax = plt.subplots(figsize=(8, 5))
box = ax.boxplot(scores, tick_labels=classes, patch_artist=True,
                 medianprops=dict(color=BAS_INK, linewidth=2))

colors = [BAS_BLUE, BAS_GREEN, BAS_AMBER, BAS_RED]
for patch, c in zip(box["boxes"], colors):
    patch.set_facecolor(c)
    patch.set_alpha(0.7)

ax.set_title("Score distribution by subject")
ax.set_ylabel("Score")
plt.tight_layout()
plt.show()
````

**Reading a box plot.** The box covers the interquartile range (Q1 to Q3). The line inside is the **median**. Whiskers extend to data within 1.5×IQR; points beyond are outliers (drawn as dots).

## 10. Multi-panel layouts — subplots

````python
rng = np.random.default_rng(7)

# A small "dashboard" — 4 different views of the same data
data = rng.normal(loc=50, scale=15, size=300)
time  = np.arange(300)
trend = np.cumsum(rng.normal(0, 1, 300))

fig, axes = plt.subplots(2, 2, figsize=(11, 7))
fig.suptitle("Mini exploratory dashboard", fontsize=14, fontweight="bold")

# (0, 0) Histogram
axes[0, 0].hist(data, bins=25, color=BAS_BLUE, edgecolor=BAS_INK)
axes[0, 0].set_title("Distribution")
axes[0, 0].set_xlabel("Value")
axes[0, 0].set_ylabel("Count")

# (0, 1) Time series
axes[0, 1].plot(time, trend, color=BAS_AMBER)
axes[0, 1].set_title("Random walk")
axes[0, 1].set_xlabel("Time")
axes[0, 1].set_ylabel("Cumulative value")

# (1, 0) Box plot
axes[1, 0].boxplot(data, vert=False, patch_artist=True,
                   boxprops=dict(facecolor=BAS_GREEN, alpha=0.6))
axes[1, 0].set_title("Box plot")
axes[1, 0].set_xlabel("Value")

# (1, 1) Scatter (data vs time)
axes[1, 1].scatter(time, data, alpha=0.5, color=BAS_RED, s=10)
axes[1, 1].set_title("Value over time")
axes[1, 1].set_xlabel("Time")
axes[1, 1].set_ylabel("Value")

plt.tight_layout()
plt.show()
````

> 💡 `axes` is a NumPy array of Axes objects — `axes[0, 0]` is top-left, `axes[1, 1]` is bottom-right. For a single row or column you get a 1-D array (`axes[0]`, `axes[1]`).

## 12. A heatmap — for matrices

````python
# Correlation matrix of 5 random features (synthetic)
rng = np.random.default_rng(0)
features = ["f1", "f2", "f3", "f4", "f5"]
X = rng.normal(size=(200, 5))
X[:, 1] = 0.8 * X[:, 0] + 0.2 * X[:, 1]    # f2 correlated with f1
X[:, 3] = -0.5 * X[:, 2] + 0.5 * X[:, 3]   # f4 anti-correlated with f3

corr = np.corrcoef(X.T)

fig, ax = plt.subplots(figsize=(6, 5))
# A *diverging* map is right here: correlations run from -1 through 0 to +1.
im = ax.imshow(corr, cmap="RdBu_r", vmin=-1, vmax=1)

ax.set_xticks(range(len(features)), features)
ax.set_yticks(range(len(features)), features)
ax.set_title("Feature correlation matrix")

# Annotate each cell
for i in range(len(features)):
    for j in range(len(features)):
        ax.text(j, i, f"{corr[i, j]:.2f}", ha="center", va="center",
                color="white" if abs(corr[i, j]) > 0.6 else BAS_INK, fontsize=9)

fig.colorbar(im, ax=ax, label="Correlation")
plt.tight_layout()
plt.show()
````

**Reading a heatmap.** Each cell is one number, encoded as colour. The diagonal is always
`1.00` (every feature correlates perfectly with itself) and the matrix is symmetric, so only one
triangle carries information. Here `f1`–`f2` come out at **+0.97** (the strong positive we built
in) and `f3`–`f4` at **−0.69** (the negative one); everything else hovers around zero.

The colormap is doing real work. Correlations *diverge* around a meaningful midpoint, so a
diverging map (`"RdBu_r"`) pinned with `vmin=-1, vmax=1` puts white at "no relationship" and
makes the sign readable at a glance. For data that only runs low→high — counts, prices, distances
— use a sequential map such as `"Blues"` or `"viridis"` instead. Never a rainbow map (`"jet"`):
it invents sharp visual edges where the data is perfectly smooth, and it is unreadable in
greyscale or with red-green colour blindness. Annotating the cells with the numbers, as above,
also lets a reader check the colours rather than trust them.

## 13. How charts mislead

Every chart is an argument. The same numbers can be drawn to support opposite arguments, and it
rarely takes bad intent — just a default nobody questioned. The classic is the **truncated
axis**: crop the y-axis and a rounding error becomes a trend.

````python
# Twelve months of a customer-satisfaction score, drawn twice from the SAME data.
months = np.arange(1, 13)
score  = np.array([7.10, 7.12, 7.15, 7.13, 7.18, 7.20,
                   7.19, 7.24, 7.26, 7.25, 7.30, 7.32])

fig, (ax_bad, ax_ok) = plt.subplots(1, 2, figsize=(11, 4), sharex=True)

# LEFT — y-axis cropped to 7.05–7.35: a +0.22 change fills the whole panel
ax_bad.plot(months, score, marker="o", color=BAS_RED, linewidth=2)
ax_bad.set_ylim(7.05, 7.35)
ax_bad.set_title("Cropped axis: 'satisfaction is soaring'")
ax_bad.set_xlabel("Month")
ax_bad.set_ylabel("Score (0–10)")

# RIGHT — the full scale the score actually lives on
ax_ok.plot(months, score, marker="o", color=BAS_BLUE, linewidth=2)
ax_ok.set_ylim(0, 10)
ax_ok.set_title(f"Full 0–10 scale: +{score[-1] - score[0]:.2f} points in a year")
ax_ok.set_xlabel("Month")
ax_ok.set_ylabel("Score (0–10)")

fig.suptitle("One dataset, two stories — the only difference is set_ylim",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()

print(f"Actual change: {score[-1] - score[0]:+.2f} points "
      f"({(score[-1] / score[0] - 1) * 100:+.1f} %)")
````

Neither panel is a lie about the *numbers* — both plot the identical twelve values. The left one
lies about their **importance**, and it does so silently, because readers judge magnitude from the
shape of the line, not from the tick labels.

A truncated axis is not automatically wrong: for a line chart of a share price around €412 you
cannot sensibly start at zero. The rules are (1) never crop the axis on a **bar** chart, where
length *is* the quantity, and (2) if you crop, say so — in the axis label, the title, or a note.

### The rest of the trap list

| Trap | Why it misleads | Do this instead |
|---|---|---|
| Truncated axis on a **bar** chart | bar length encodes the value, so cropping inflates differences | start bar axes at zero, always (`ax.set_ylim(0, …)`) |
| **Unsorted** category bars | the reader must search for the ranking, and may miss it | sort by value unless the categories have their own order |
| **Dual y-axes** (`ax.twinx()`) | two independent scales can be slid until *any* two series look coupled | two stacked panels sharing the x-axis: `plt.subplots(2, 1, sharex=True)` |
| **Pie chart** with many slices | angles compare badly; small slices and their labels collide | a sorted `ax.barh` of the shares |
| **Rainbow** colormap (`"jet"`) for ordered data | invents sharp edges where the data is smooth; fails in greyscale and for colour-blind readers | sequential `"Blues"` / `"viridis"`; diverging `"RdBu_r"` around a midpoint |
| **Hiding sample size** | a bar built on 3 observations looks exactly like one built on 3 000 | put *n* in the tick label, or overlay the raw points |
| **3-D effects, cut-out shapes** | perspective distorts area; decoration competes with data | plain 2-D, and delete anything that carries no information |

One check catches most of these: **say out loud the sentence your chart makes a reader believe,
then ask whether the data supports that sentence.** "Satisfaction is soaring" fails the test;
"satisfaction rose 0.2 points, which is barely anything" passes. In Notebook 11 you will run this
check on charts of real data, where the sentence is not one you already know the answer to.

## 14. Saving figures

A figure you cannot get out of the notebook is not a deliverable. The Figure object saves itself:

```python
fig, ax = plt.subplots()
ax.plot(x, y)
fig.savefig("revenue.png", dpi=200, bbox_inches="tight")   # before plt.show()!
```

> 📁 The line above is deliberately *not* executed anywhere in this notebook — these materials
> never write files, so they behave identically on your laptop and in a throwaway Colab session.
> Copy it into your own work and it will write the file next to your notebook.

Common formats:

| Extension | When to use                                          |
|-----------|------------------------------------------------------|
| `.png`    | reports, web, default raster                          |
| `.pdf`    | papers, slides, infinitely scalable vectors           |
| `.svg`    | web vectors / further editing in Inkscape, Illustrator |

`bbox_inches="tight"` crops out unused whitespace; `dpi=200` (or 300) gives a sharp result. Save
*before* `plt.show()`: in many backends `show()` empties the figure, and you get a blank file.

## 15. Common technical pitfalls

| Pitfall                                       | Symptom                                | Fix |
|-----------------------------------------------|----------------------------------------|-----|
| Figures look squished / overlap labels        | crowded titles / labels                | call `plt.tight_layout()` before `plt.show()` |
| Calling `plt.plot` multiple times in same cell | unintended cumulative figure           | use `fig, ax = plt.subplots()` for each plot |
| Wrong axis order on heatmap                   | matrix looks rotated/mirrored          | remember `imshow` plots `data[row][col]`, with row 0 on top |
| Colorbars on subplots                         | the colorbar covers a neighbour        | pass `ax=` to `fig.colorbar(im, ax=ax)` |
| Bad colour choices                            | unreadable / non-colourblind-friendly  | prefer perceptual maps like `"Blues"`, `"viridis"` or `"cividis"` |
| `fig.savefig` after `plt.show()`              | the saved file is blank                | save first, show second |
| One colour per bar in a single series          | colour implies a grouping that isn't there | one colour for the series; colour only to *encode* something |

## 🧠 Key takeaways

1. Use the **Figure / Axes** (`fig, ax = plt.subplots()`) style for everything beyond a quick one-off.
2. Always label your axes and give the plot a title — a chart without context is a riddle.
3. Match the **chart type** to the question:
   - line for *trend over time*,
   - bar for *comparing categories*,
   - histogram / box for *distributions*,
   - scatter for *relationships*,
   - heatmap for *matrices*.
4. Build multi-panel layouts with `plt.subplots(rows, cols)` for "data dashboards".
5. Annotate the punchline — arrows, vertical lines, callouts — and let the **title state the
   finding**, not the axis names.
6. Save with `fig.savefig("name.png", dpi=200, bbox_inches="tight")`, *before* `plt.show()`.
7. Stick to a small, consistent palette for categories (here: the `BAS_*` brand colours) and a perceptually uniform colormap (`"Blues"`, `"viridis"`) for continuous values — never a rainbow map for ordered data.
8. **A chart is an argument.** Zero-based bar axes, sorted categories, no dual y-axes, *n* stated.
   Say the sentence your chart makes a reader believe, and check the data supports it.
